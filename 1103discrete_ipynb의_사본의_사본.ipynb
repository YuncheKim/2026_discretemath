```python
graph_n = {
    '홍길동': ['김윤채', '하지민', '이지은', '조연주', '허남재'], #추가생성,인싸 홍길동씨


    '김윤채': ['김근영', '김성연', '조연주', '목은혜', '김다경'],
    '이지은': ['노다운', '이재은', '최시훈', '유채영', '한진주'],
    '최연희': ['하지민', '신지민', '백선우', '박은영', '허남재'],
    '양다영': ['김희진', '조수아', '김지은', '오수현', '고가은'],
    '박건우': ['이준서', '정우진', '최지훈', '전민성', '박지후']

}
```

김윤채| 김근영 김성연 조연주 목은혜 김다경

이지은| 노다운 이재은 최시훈 유채영 한진주

최연희| 하지민 신지민 백선우 박은영 허남재

양다영| 김희진 조수아 김지은 오수현 고가은

박건우| 이준서 정우진 최지훈 전민성 박지후

```python
def dfs(graph_n, v, visited):
    visited.add(v)
    print(v, end=' ')
    for neighbor in graph_n.get(v, []):
        if neighbor not in visited:
            dfs(graph_n, neighbor, visited)

def DFS(graph_n):
    visited = set()
    for v in graph_n:
        if v not in visited:
            dfs(graph_n, v, visited)
    print()

def bfs(graph_n, v, visited):
    queue = [v]
    visited.add(v)
    while queue:
        u = queue.pop(0)
        print(u, end=' ')
        for neighbor in graph_n.get(u, []):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

def BFS(graph_n):
    visited = set()
    for v in graph_n:
        if v not in visited:
            bfs(graph_n, v, visited)
    print()

print("DFS탐색 순서:")
DFS(graph_n)
print("BFS탐색 순서:")
BFS(graph_n)
```

```python
print("DFS탐색 순서:")
dfs(graph_n, '홍길동', set())
```

```python
print("BFS탐색 순서:")
bfs(graph_n, '홍길동', set())
```

```python
def recommend_friend(graph_n, start):
    visited = set()
    queue = [(start, 0)]
    print(f"\n{start}의 친구 추천 (2단계):")

    while queue:
        current, depth = queue.pop(0)
        if current not in visited:
            visited.add(current)
            if depth == 2:
                print("-", current)
            if depth < 2:
                for neighbor in graph_n.get(current, []):
                    queue.append((neighbor, depth + 1))

recommend_friend(graph_n, '홍길동')
```
